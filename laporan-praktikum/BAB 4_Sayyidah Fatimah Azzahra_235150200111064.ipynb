### Soal 1
**1. Implementasikan metode normalisasi Z-score dengan cara membuat fungsi bernama score.**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = pd.read_csv('dataset/Iris.csv')
data.info()
```

```python
data = data.drop(columns=['Id'])
```

```python
def zscore(data):
    return (data - data.mean()) / data.std()
```

```python
fitur_numerik = data.select_dtypes(include=['int64', 'float64']).columns

normalized_data = data[fitur_numerik].apply(zscore)
print(normalized_data)
print(data)
```

#### Penjelasan

Setiap nilai pada kolom-kolom pada dataset dinormalisasi dengan menggunakan fungsi zscore, yang bekerja dengan cara mengurangi nilai dengan rata-rata dan membaginya dengan standar deviasi.

### Soal 2

**2. Normalisasikan DataFrame data_imputasi menggunakan fungsi zscore. Simpan hasilnya pada
DataFrame bernama data_zscore.**

```python
def imputasi(df_input) :
    list_columns = df_input.columns #mendapatkan daftar kolom pada DataFrame
    class_column = list_columns[-1] #kolom terakhir merupakan kolom kelas
    for column in list_columns [:-1]:
        # Penggantian nilai NaN dilakukan per kolom
        df_input[column] = df_input[column].fillna(
            df_input.groupby(class_column)[column].transform('mean'))
    return df_input

data_imputasi = imputasi(data)
data_imputasi.info()
```

```python
data_zscore = data_imputasi[fitur_numerik].apply(zscore)
data_zscore
```

#### Penjelasan

Sama halnya seperti soal sebelumnya, saya mengimplementasikan fungsi zcore ini pada dataset, namun setelah dilakukan imputasi dengan mean.

### Soal 3

**3. Jelaskan perbedaan hasil normalisasi MinMax dan Z-Score. Petunjuk: cek nilai rentang, rata-rata dan simpangan baku dari data_zscore.**

```python
def minmax_normalization(data):
    min_val = min(data)
    max_val = max(data)
    normalized_data = [(x - min_val) / (max_val - min_val) for x in data]
    return normalized_data

data_minmax = data_imputasi[fitur_numerik].apply(minmax_normalization)
data_minmax.head()
```

```python
print(data_zscore.describe())
print(data_minmax.describe())
```

#### Penjelasan

Perbedaan hasil normalisasi MinMax dan Z-Score:

1. Rentang nilai: Z-Score berkisar -2.4 hingga 3.1, sementara MinMax selalu 0 sampai 1.

2. Rata-rata: Z-Score mendekati 0 untuk semua fitur, MinMax bervariasi antara 0.42-0.47.

3. Simpangan baku: Z-Score selalu 1, MinMax bervariasi antara 0.18-0.32.

4. Interpretasi: Z-Score menunjukkan jarak nilai dari rata-rata dalam standar deviasi, MinMax menunjukkan posisi relatif nilai dalam rentang aslinya.

5. Sensitivitas terhadap outlier: Z-Score lebih tahan, MinMax lebih sensitif karena bergantung pada nilai ekstrem.

Kedua metode memiliki kelebihan masing-masing tergantung kebutuhan analisis dan karakteristik data. Misalnya, Z-Score lebih tahan terhadap outlier, sedangkan MinMax lebih sensitif karena bergantung pada nilai ekstrem.
